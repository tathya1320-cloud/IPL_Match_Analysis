```python
# =============================
# # IPL Match Data Analysis
# =============================
# **Project Overview:**
# This project analyzes IPL match data to understand team performance,
# toss impact, and win patterns. Key metrics include total matches,
# team-wise wins, win percentages, toss advantage, and visualizations.
# Tools: Python, Pandas, Matplotlib, Google Colab (Jupyter Notebook)

# -----------------------------
# 1️⃣ Import Libraries
# -----------------------------
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# 2️⃣ Load Data
# -----------------------------
data = {
    "season": [2020, 2020, 2021, 2021, 2022, 2022, 2023, 2023],
    "team1": ["MI", "CSK", "RCB", "KKR", "MI", "CSK", "GT", "RR"],
    "team2": ["CSK", "MI", "KKR", "RCB", "RCB", "RR", "MI", "CSK"],
    "winner": ["MI", "CSK", "RCB", "KKR", "MI", "CSK", "GT", "RR"],
    "toss_winner": ["CSK", "MI", "RCB", "KKR", "MI", "RR", "GT", "CSK"]
}

df = pd.DataFrame(data)
df

# -----------------------------
# 3️⃣ Analysis
# -----------------------------

# Total Matches
total_matches = len(df)
print("Total Matches:", total_matches)

# Team-wise Wins
win_counts = df["winner"].value_counts()
print("\nTeam-wise Wins:\n", win_counts)

# Win Percentage
win_percentage = (win_counts / total_matches) * 100
print("\nWin Percentage (%):\n", win_percentage)

# Toss Advantage Percentage
toss_advantage = (len(df[df["toss_winner"] == df["winner"]]) / total_matches) * 100
print("\nToss Advantage Percentage:", toss_advantage, "%")

# -----------------------------
# 4️⃣ Visualizations
# -----------------------------

# Most Successful Teams (Count)
win_counts.plot(kind="bar", color='skyblue')
plt.title("Most Successful IPL Teams")
plt.xlabel("Team")
plt.ylabel("Number of Wins")
plt.show()

# Win Percentage
win_percentage.plot(kind="bar", color='orange')
plt.title("Win Percentage of IPL Teams")
plt.xlabel("Team")
plt.ylabel("Win Percentage")
plt.show()

# -----------------------------
# 5️⃣ Insights
# -----------------------------
# MI and CSK are the most successful teams in this dataset.
# Toss advantage exists but is not dominant (~50%).
# No single team dominates consistently.
# Visualizations clearly show win distribution and percentage.
```
